# Multimodal Data Preprocessing: Identity and Product Recommendation

A person must pass a **face check**, then a **voice check**, before they see a product prediction.
This notebook covers every step: merging the customer tables, processing face and voice samples, training the three models, and simulating real and unauthorized attempts.

**Run order:** open from the repository root or `notebooks/`, then run all cells. The code lives in `src/`, so this notebook and the command-line app share one implementation.

## 2. Cleaning and merge

**Cleaning**
- Transactions: `purchase_date` becomes a datetime. The 10 missing ratings are kept as missing in `customer_rating`, flagged in `rating_missing`, and median-filled only for rating history.
- Social profiles: 5 exact duplicate rows are removed. `review_sentiment` becomes -1, 0 or 1.

**Join key:** `A178` in social profiles and `178` in transactions are treated as the same customer, because the IDs share one numeric range and no other key exists. The files do not state this, so it is an assumption.

**Why not join directly?** 41 customers have several social rows, so a direct join copies transactions. The next cell shows the effect. The fix is to aggregate social data to one row per customer first, then join with `validate='many_to_one'`, which raises an error if a row would be duplicated.

**Why an inner join?** The model predicts from social behaviour, so a transaction with no social profile has no usable input. Those transactions are dropped from the merged table, but they still count toward customer purchase history.

```python
tx, tx_log = tabular.clean_transactions(tx_raw)
social, social_log = tabular.clean_social(social_raw)
print('Transactions:', tx_log)
print('Social:', social_log)

direct = tx.merge(social, on='customer_id')
print(f'\nDirect join: {len(direct)} rows from {direct.transaction_id.nunique()} transactions '
      f'({len(direct) - direct.transaction_id.nunique()} copied rows)')

merged, logs = tabular.build_dataset()
checks = logs['merge']
display(pd.Series(checks, name='value').to_frame())
assert merged['transaction_id'].is_unique
assert len(merged) == merged.groupby('customer_id').size().sum()
assert merged[tabular.TARGET].notna().all()
assert merged['interest_mean'].between(1, 5).all() and merged['engagement_mean'].between(50, 99).all()
print('Post-merge checks passed: unique transaction IDs, no missing target, scores in range.')
display(merged.head())
```

### Feature engineering, and what is left out

| Group | Features | Source |
|---|---|---|
| Social behaviour | mean and max engagement, mean and max interest, mean sentiment, records, platforms used, share per platform, primary platform | All social rows per customer |
| Purchase history | prior purchase count, average spend, average rating, days since last purchase, last category | **Earlier** transactions only |
| Calendar | month, day of week | Visit date |

`purchase_amount` and `customer_rating` from the current transaction are **excluded from the model**, because they do not exist until after the purchase being predicted. History is computed with a shift, so a row never sees its own target.

The social table has no dates, so its aggregates may include activity after some purchases. This is a limitation that the data cannot resolve.

```python
fig = eda.plot_distributions(tx_raw, social_raw); plt.show()
fig = eda.plot_outliers(tx_raw, social_raw); plt.show()
display(eda.iqr_outlier_counts(tx_raw, social_raw).round(2))
```

**Reading the distributions and outliers.** All four variables spread broadly across their ranges with no strong peak. Amounts run from 51 to 495 with a mean of 281 and median of 274, so there is little skew. The IQR rule flags **no outliers**, so nothing is removed or capped.

```python
fig = eda.plot_correlations(merged); plt.show()
fig = eda.plot_target(merged); plt.show()

from scipy.stats import chi2_contingency, f_oneway
for col in ['interest_mean', 'engagement_mean', 'sentiment_mean']:
    p = f_oneway(*[g[col] for _, g in merged.groupby('product_category')]).pvalue
    print(f'ANOVA, {col} across products: p = {p:.3f}')
p = chi2_contingency(pd.crosstab(merged['primary_platform'], merged['product_category']))[1]
print(f'Chi-square, primary platform by product: p = {p:.3f}')
```

**Reading the relationships.** No pair of variables correlates beyond about 0.25 in absolute value, so there is no redundant feature to drop. Classes are reasonably balanced, from 20 to 28 rows each.

The tests matter most for what follows: none of the social features differs significantly across products at the 5% level. Interest comes closest (p ≈ 0.07). **Expect a weak product model:** the features carry little signal about which category someone buys.

- **Grayscale and RGB histograms**: tone and colour

## 5. Voice samples

Each member records *Yes, approve* and *Confirm transaction*.

**How to read the plots.** On a waveform, bursts are spoken syllables and flat stretches are pauses. On a mel spectrogram, time runs left to right, pitch runs bottom to top, and bright bands show where a voice concentrates its energy. Those bands depend partly on the speaker's vocal tract, which is why they help tell people apart. The printed lines under the plots give each clip's measured reading.

```python
clips = audio.list_enrolled(DATA_ROOT)
audio_members = sorted({c['member'] for c in clips})
fig, axes = plt.subplots(len(audio_members), 4, figsize=(14, 2.3 * len(audio_members)), squeeze=False)
readings = []
for r, member in enumerate(audio_members):
    for c, phrase in enumerate(PHRASES):
        path = next(x['path'] for x in clips if x['member'] == member and x['phrase'] == phrase)
        y = audio.load_audio(path)
        info = audio.describe_clip(y)
        readings.append({'member': member, 'phrase': phrase, **{k: round(v, 1) for k, v in info.items()}})
        t = np.arange(len(y)) / SAMPLE_RATE
        wave, spec = axes[r, 2 * c], axes[r, 2 * c + 1]
        wave.plot(t, y, color=eda.BLUE, linewidth=0.5)
        wave.set(title=f'{member}: {phrase}', xlabel='Seconds', ylabel='Amplitude', ylim=(-1.05, 1.05))
        spec.imshow(audio.mel_spectrogram_db(y), origin='lower', aspect='auto', cmap='magma', extent=[0, t[-1], 0, audio.N_MELS])
        spec.set(title='Mel spectrogram', xlabel='Seconds', ylabel='Mel band')
        spec.grid(False)
fig.tight_layout(); plt.show()

readings = pd.DataFrame(readings)
display(readings)
for row in readings.itertuples():
    quality = 'clean, with clear pauses' if row.loud_vs_quiet_db >= 20 else 'noisy, as speech barely rises above the background'
    clipping = 'no meaningful clipping' if row.clipped_pct < 0.5 else f'{row.clipped_pct}% of samples clipped, so re-record quieter'
    print(f'{row.member} / {row.phrase}: {row.duration_s}s, median pitch {row.median_pitch_hz} Hz, '
          f'{row.energy_below_4khz_pct}% of energy below 4 kHz; {quality}; {clipping}.')
```

### Augmentation and features

Every clip gets **five augmentations**: pitch ±1 semitone, time stretch at 0.9× and 1.1×, and background noise at 20 dB SNR. Pitch shift is kept small on purpose, because a large shift starts to imitate a different person.

Features saved to `audio_features.csv`: mean and spread of 13 MFCCs, spread of their deltas, spectral roll-off, spectral centroid, RMS energy, zero-crossing rate and pitch.

```python
first = audio.load_audio(clips[0]['path'])
variants = audio.augment(first)
fig, axes = plt.subplots(2, 3, figsize=(12, 4.2), sharey=True)
for ax, (name, clip) in zip(axes.flat, variants.items()):
    ax.plot(np.arange(len(clip)) / SAMPLE_RATE, clip, color=eda.BLUE, linewidth=0.4)
    ax.set(title=name, xlabel='Seconds', ylabel='Amplitude')
fig.suptitle(f"Augmentations of {clips[0]['path'].name}", x=0.02, ha='left', fontweight='bold')
fig.tight_layout(); plt.show()

audio_features = audio.build_audio_features(DATA_ROOT, DATA_SOURCE)
print(f'audio_features.csv: {audio_features.shape[0]} rows, {audio_features.shape[1]} columns')
print(audio_features.groupby('member').size().to_dict())
display(audio_features[['audio_id', 'member', 'augmentation', 'mfcc_1_mean', 'mfcc_2_mean', 'spectral_rolloff_mean', 'energy_rms_mean', 'f0_mean']].head(6).round(2))
```

## 6. Models and evaluation

**How each score is kept honest**

| Model | Test design | What it prevents |
|---|---|---|
| Face | Train on two expressions, test on the third | Augmented copies of a test photo appearing in training |
| Voice | Train on one phrase, test on the other | The same, and it tests the speaker rather than the words |
| Product | 5-fold CV repeated 5 times, each customer in one fold only; plus a time-ordered holdout and a label-shuffle test | A customer's profile appearing on both sides; chance results looking meaningful |

Candidates are logistic regression and random forest for face and voice, plus gradient boosting and a class-frequency baseline for product. Metrics are accuracy, macro F1 and log loss.

**Rejecting strangers.** A normal classifier always names somebody, often with high confidence. So face and voice must also pass a similarity check against that person's enrolled samples, using a separate PCA and LDA embedding. Strangers in `unauthorized/` tune this threshold, with each stranger held out in turn.

The next cell retrains everything and saves the models. It takes a few minutes because of the label-shuffle test.

```python
run_pipeline.main()
metrics = json.loads((REPORTS / 'metrics.json').read_text())
```

```python
rows = []
for modality in ('face', 'voice'):
    for name, r in metrics[modality]['cv'].items():
        rows.append({'task': modality, 'model': name, 'selected': name == metrics[modality]['selected_model'],
                     'accuracy': r['accuracy'], 'f1_macro': r['f1_macro'], 'log_loss': r['log_loss']})
for name, r in metrics['product']['cv_customer_grouped'].items():
    rows.append({'task': 'product', 'model': name, 'selected': name == metrics['product']['selected_model'],
                 'accuracy': r['accuracy'], 'f1_macro': r['f1_macro'], 'log_loss': r['log_loss']})
display(pd.DataFrame(rows).round(3))

print('Face and voice data source:', metrics['face_voice_data_source'])
display(pd.DataFrame({m: metrics[m]['open_set'] for m in ('face', 'voice')}).round(3))

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, name in zip(axes, ['confusion_face.png', 'confusion_voice.png', 'confusion_product.png']):
    ax.imshow(plt.imread(FIGURES / name)); ax.axis('off')
fig.tight_layout(); plt.show()
```

```python
p = metrics['product']
best, base = p['cv_customer_grouped'][p['selected_model']], p['cv_customer_grouped']['baseline_class_frequency']
perm, hold = p['permutation_test'], p['time_holdout']
print(f"Product model: {p['selected_model']} on {p['n_rows']} transactions from {p['n_customers']} customers")
print(f"  Customer-grouped CV : accuracy {best['accuracy']:.3f} vs baseline {base['accuracy']:.3f}")
print(f"                        log loss {best['log_loss']:.3f} vs baseline {base['log_loss']:.3f} (lower is better)")
print(f"  Label-shuffle test  : shuffled labels average {perm['null_mean_accuracy']:.3f}; p = {perm['p_value']:.3f}")
print(f"  Time holdout        : accuracy {hold[p['selected_model']]['accuracy']:.3f} vs baseline "
      f"{hold['baseline_class_frequency']['accuracy']:.3f}, on only {hold['test_rows']} rows from {hold['test_starts']}")
```

**Reading the product result.** The model's accuracy is about 30%, against about 22% for always guessing the most common class. That gap is **not reliable evidence of skill**:
- It is not significant at the 5% level in the label-shuffle test.
- Its log loss is slightly worse than the baseline, so its probabilities are no better than the class frequencies.
- The time holdout has only 24 rows, so it cannot settle the question.

This agrees with the EDA, and the likely cause is the data, not the choice of algorithm. With 117 rows across 61 customers, and only 56 rows having any earlier purchase, more tuning would mostly fit noise. Accuracy reported from a random row split, or with purchase amount and rating included, should be treated with caution for the reasons above.

## 7. Multimodal decision logic and simulation

This follows the assignment's diagram step by step:

1. **Face model** names an enrolled member, or the attempt is denied.
2. **Product model** runs for that member's customer record. The result is held back.
3. **Voice model** must accept the voice **and** name the same person as the face. Otherwise, the held result is discarded.
4. **Display** happens only after both checks pass.

The same-person rule matters. With two independent checks, one member's face and another member's voice would pass.

`data/members.json` links each member to a customer record for the demo, because the source data has no such link. The scripted cases use samples that were not used in training.

```python
system = app.System()
app.banner(system)
outcomes = {name: app.run_scenario(system, name) for name in app.SCENARIOS}
expected = {'authorized': True, 'unknown_face': False, 'unknown_voice': False, 'voice_of_other_member': False}
display(pd.DataFrame([{'scenario': k, 'expected_access': expected[k], 'access_granted': v['granted'],
                       'stopped_at': v.get('stopped_at', '-'), 'reason': v.get('reason', '-')} for k, v in outcomes.items()]))
```

## 8. Conclusions and limits

- **Merge:** aggregating social data first keeps one row per transaction and avoids copied rows. The cost is losing 33 of 150 transactions that have no social profile.
- **Product model:** it runs correctly, but the data does not support confident predictions. More customers and timestamped social activity would be the most useful improvements.
- **Face and voice:** HOG and MFCC features are simple and easy to explain, but they are weaker than deep embeddings. A few samples per person cannot establish real-world security, and a photo or recording could fool this system because it has no liveness check.
- **Voice check:** it identifies the speaker, not the words, so a different phrase from the right person is accepted.
